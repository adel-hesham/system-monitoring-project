## Mimic CPU-Z info as much as you can

```python
import psutil
import platform
import subprocess
import time

##cpu cache

cpu_cache= subprocess.check_output(
    "lscpu | grep '^L[1:3]*'",
    shell=True
).decode().strip()
print(cpu_cache)


#operanting system
os = subprocess.run(
    ["cat /etc/os-release | grep PRETTY_NAME | cut -f2 -d ="],
    text=True,
    shell=True,
    capture_output=True
)
print(os.stdout.strip())



####cpu  info#####
cpu_joule1 = subprocess.run(
    ["sudo", "-S", "cat", "/sys/class/powercap/intel-rapl:0/energy_uj"],
    input="1234\n",
    text=True,
    capture_output=True
)

time.sleep(1)

cpu_joule2 = subprocess.run(
    ["sudo", "-S", "cat", "/sys/class/powercap/intel-rapl:0/energy_uj"],
    input="1234\n",
    text=True,
    capture_output=True
)


cpu_power = (int(cpu_joule2.stdout) - int(cpu_joule1.stdout)) / 1_000_000

print(f"CPU Power: {cpu_power:.2f} W")

# cpu name
cpu_name = subprocess.check_output(
    "lscpu | grep 'Model name'",
    shell=True
).decode().split(":")[1].strip()
print(cpu_name)


# cpu arch

cpu_arch=platform.processor()
print(cpu_arch)

#cpu temp 

cpu_temp=int(subprocess.check_output("cat /sys/class/hwmon/hwmon5/temp1_input", shell=True))/1000
print(f"{cpu_temp}°C")
#cpu voltage

cpu_voltage=int(subprocess.check_output("cat /sys/class/hwmon/hwmon2/in0_input", shell=True))/1000  
print(f"{cpu_voltage} V")

# logical=True includes hyperthreaded virtual cores
logical_cores  = psutil.cpu_count(logical=True)
physical_cores = psutil.cpu_count(logical=False)
print(f"Physical cores : {physical_cores}")
print(f"Logical cores  : {logical_cores}")
print()

# cpu frequency in MHz
freq = psutil.cpu_freq()
if freq:
    print(f"CPU frequency  : {freq.current:.0f} MHz  (min: {freq.min:.0f}, max: {freq.max:.0f})")
```
